```python
q3 = df['Study hrs'].quantile(0.75)
q3
```

```python
q1 = df['Study hrs'].quantile(0.25)
q1
```

```python
iqr = q3 - q1
iqr
```

```python
upper = q3 + (1.5 * iqr)
upper
```

```python
lower = q1 - (1.5 * iqr)
lower
```

```python
lower = np.float64(0)
lower
```

```python
df2 = df[(df['Study hrs'] <= upper) & (df['Study hrs'] >= lower)]
df2
```

```python
df.loc[df['Study hrs'] > upper, 'Study hrs'] = (upper // 1)
df
```
