```python
import spacy
```

```python
nlp = spacy.load('en_core_web_sm')
```

#### Tokenization
##### Breaking a text into smaller parts (words or tokens)
Why :
##### Syatem/Computers can not understand full sentences of paragraph directly, so spliting them into words

```python
text = '''Dr. Smith lives in the U.S. and works at New York University. 
He specializes in A.I. and has published over 3.14 papers every year. 
He often travels to Washington D.C. for conferences, workshops, etc. 
Recently, he received an award from the A.I. Society.'''
```

```python
doc = nlp(text)
```

```python
doc.sents
```

```python
for sen in doc.sents:
    print(sen)
```

#### Identify POS(Parts of Speech)

```python
text = 'How are you?'
doc = nlp(text)
for token in doc:
    print(token, ':', token.pos_)
```

#### Stemming / Lemmatization

##### Stemming
Cuts words to their root form (not always meaningful)

##### Lemmatization
Converts words into meaningful base form (dictionary form)

```python
import nltk
```

```python
from nltk.stem import PorterStemmer
obj = PorterStemmer()
```

```python
text = "I am meeting him tomorrow at the meeting"
str1 = text.split(" ")
str1
```

```python
for s in str1:
    print(obj.stem(s))
```

```python
doc = nlp(text)
for token in doc:
    print(token, '-', token.lemma_, '-', token.pos_)
```

#### Stop word removal
1. Removing common words  that don't add much meaning
2. Reduce noise and improves model performance.

```python
text = 'India is a developing country. Mumbai is economical capital. Mr. Fadanvis is Chief Minister'
stop_words = nlp.Defaults.stop_words
result = []
doc = nlp(text)
for token in doc:
    token = str(token)
    token = token.lower()
    if token not in stop_words:
        result.append(token)
result
```

#### Entity recognition
1. Identifing important real-world entities in text
2. Information extraction

```python
for token in doc.ents:
    print(token, '-', token.label_, '-', spacy.explain(token.label_))
```
