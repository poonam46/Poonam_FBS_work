2. One Hot Encoding

```python
#Pandas
data = pd.get_dummies(df['AirBags'])
data
```

```python
#Pandas
data = pd.get_dummies(df['AirBags'], dtype =int)
data
```

```python
#sklearn
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe = OneHotEncoder()
ohe.fit_transform(df[['AirBags']]).toarray()
```

```python
ohe = OneHotEncoder(sparse_output = False)
d  = df['AirBags'].unique()
```

```python
d.sort()
d
```

```python
data = pd.DataFrame(ohe.fit_transform(df[['AirBags']]), columns = d, dtype=int)
data.head()
```

```python
df2 = df.drop('AirBags', axis=1)
```

```python
df2.head()
```

```python
df2 = pd.concat([df2, data], axis=1)
df2.head()
```
