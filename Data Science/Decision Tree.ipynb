```python
# Divide the input and output
X = data.drop('Species', axis =1)
y = data['Species']
```

```python
# Here no balancing of data required
```

```python
# No scaling required
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 30)
```

```python
# Decide best hyperparameter using GridSearchCV
```

```python
# Build the model with best hyperparameter
```

```python
from sklearn.tree import DecisionTreeClassifier
model = DecisionTreeClassifier()
model.fit(X_train, y_train)
```

```python
from sklearn.model_selection import cross_validate
result = cross_validate(model, X_train, y_train)
result
```

```python
print("Validation Score : ", result['test_score'].mean())
```

```python
train_pred = model.predict(X_train)
test_pred = model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
print('Training Accuracy Score : ', accuracy_score(y_train, train_pred))
print('Testing Accuracy Score : ', accuracy_score(y_test, test_pred))
```

```python
from sklearn.tree import plot_tree
plt.figure(figsize = (7, 10))
plot_tree(model, class_names = data['Species'].unique(), feature_names= X.columns, filled=True)
plt.show()
```
