```python
import numpy as np
from scipy import stats
```

```python
#sample data
data2 = [70, 65, 74, 62, 66, 78, 68, 71]
data2
```

```python
data = np.random.randint(50, 90, 10)
data
```

```python
#null hypothesis
mu = 70
```

```python
#Perform one sample: t-test, chi-test, z-test

#t - test : sample value < 30 (numeric data)
#z - test : sample value >= 30 (numeric data)
#chi - test : sample - categorical (text data) -> (good, bad, average)
#statistics : 'How far is my sample result from what I expected (H0)'

statistics, p_value = stats.ttest_1samp(data, mu)
```

```python
print(f'Satistics : {statistics:.4f}')
print(f'P-value : {p_value:.2f}')
```

```python
#decision
alpha = 0.05
if(p_value > alpha):
    print('Failed to reject the null hypothesis (H0)')
else:
    print('Reject the null hypothesis (H0)')
```
