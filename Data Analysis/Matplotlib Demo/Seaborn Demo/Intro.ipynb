### Seaborn
1. Python library used for data visualization
2. Build on top of the matplotlib python library
3. High quality graphs
4. Use for statistical graphics

```python
import seaborn as sns
```

```python
data = sns.load_dataset('tips')    # Load the data from github
data.head()
```

```python
sns.lineplot(x = 'total_bill' , y = 'tip', data = data, hue= 'sex')
```

```python
sns.barplot(x = 'day', y = 'total_bill', data= data)#hue = 'day')
```

```python
sns.histplot(data['total_bill'], kde = True)   # kde = Kernel Density Estimation
```

```python
sns.scatterplot(x = 'total_bill', y = 'tip', hue = 'sex', data = data)
```

### Heat Map
1. Graphical reperesentation of data when values are represented using colors
2. It required numeric data
3. Dark / warm color -> higher value
4. Lighter / cooler color -> lower value
5. Easy to identify correlation
6. Works well with large dataset
7. Helps to feature engineering
