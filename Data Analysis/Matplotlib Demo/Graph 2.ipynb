```python
from matplotlib import pyplot as plt
```

### Pie Plot
1. Showing percentage distribution
2. Part-to-whole comparison
3. Best when categories are limited(<= 5)
4. Parts will be mentioned anti-clock wise

```python
sub = ['Python','MySQL','Data Analysis','Data Science', 'Web Development','Frameworks']
marks = [80, 70, 85, 90, 60, 45]
explode = [0.03, 0, 0.1, 0, 0.05, 0]
colors = ['c', 'b','r','y','g','m']
plt.pie(marks, labels = sub, explode = explode, colors = colors, autopct = '%0.2f%%')
plt.show()
```

### Box Plot
1. Summary of distribution
2. Detecting Outliers
3. Very useful in exploratory data analysis (EDA)

1. q1 = 25%
2. q3 = 75%

IQR (Interquartile Range) = q3 - q1

1. minimum = q1 - 1.5 * IQR
2. maximum = q3 + 1.5 * IQR

```python
data = [1, 7, 8, 9, 10, 11, 17]
plt.boxplot(data)
plt.show()
```
