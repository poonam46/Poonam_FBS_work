#### Bar Plot
1. Comparing Categories
2. Showing frequency or count
3. Product sales comparison
4. Identify highest / lowest values easily

```python
import numpy as np
```

```python
drinks = ['Tea','Hot Coffee', 'Cold Coffee', 'Juice']
summer = [70, 50, 80, 60]
winter = [85, 70, 30, 50]
drinks_index = np.arange(len(drinks))
width = 0.3
```

```python
plt.bar(drinks_index, summer, width, color='purple', label= 'summer')
plt.bar(drinks_index+width+0.1, winter, width, color = 'red', label = 'winter')
plt.xlabel('Drinks')
plt.ylabel('Count')
plt.legend()
plt.show()
```

```python
days = np.arange(1, 31)
temp_delhi = np.random.randint(30, 50, 30)
temp_mumbai = np.random.randint(30, 50, 30)
```
