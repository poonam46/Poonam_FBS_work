```python
from matplotlib import pyplot as plt
```

```python
import numpy as np
```

```python
over = np.arange(1,21)
#print(over)
india = np.random.randint(0, 30, 20)
england = np.random.randint(0, 30, 20)
over_index =  np.arange(len(over))
width = 0.3
```

```python
plt.figure(figsize = (12,6))
bar1 = plt.bar(over_index-width/2, india , width, color= 'blue', label='India')
bar2 = plt.bar(over_index+width/2, england, width, color = 'red', label='England')
plt.title('India vs England Runs Per Over')
plt.xlabel('Overs')
plt.ylabel('Runs')
plt.legend()
plt.xticks(over_index,over)
#plt.tight_layout()
plt.xlim(-0.5,20.5)
plt.margins(x=0)
plt.bar_label(bar1)
plt.bar_label(bar2)
plt.show()
```

```python
plt.figure(figsize = (12,6))
plt.plot(over, india , marker = 'o', color = 'blue', label = 'India' )
plt.plot(over, england, marker= 'o', color = 'red', label = 'England')
plt.title('India vs England Runs Per Over')
plt.xlabel('Overs')
plt.ylabel('Runs')
plt.legend()
plt.xticks(np.arange(1,21))
plt.show()
```
