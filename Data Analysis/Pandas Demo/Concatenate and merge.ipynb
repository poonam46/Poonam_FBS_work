```python
import pandas as pd
```

```python
emp = pd.DataFrame({'ID' : [101,102,103,104,105], 'Name' : ['Kiran','Om','Poonam','Anushka','Vikram'],
                    'Dept_ID' : [1, 2, 1, 3, 2]})
emp
```

```python
dept = pd.DataFrame({'ID' : [1,2,3,4] , 'DEPT_NAME' : ['Developer','Tester', 'HR', 'Admin']})
dept
```

```python
pd.concat([emp, dept])
```

```python
pd.concat([emp, dept], axis = 1)
```

```python
emp.loc[emp['Name'] == 'Vikram', 'Dept_ID'] = 5
```

```python
emp
```

```python
pd.merge(emp, dept, left_on = 'Dept_ID', right_on = 'ID', how = 'inner')
```

```python
pd.merge(emp, dept, left_on = 'Dept_ID', right_on = 'ID', how = 'left')
```

```python
pd.merge(emp, dept, left_on = 'Dept_ID' , right_on = 'ID' , how = 'right')
```

```python
pd.merge(emp, dept, left_on = 'Dept_ID' , right_on = 'ID' , how = 'outer')
```
