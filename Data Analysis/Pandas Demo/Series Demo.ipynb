#### Intro
pd.Series()
1. Series is a data structure in pandas.
2. It is 1 dimensional data structure.
3. S is capital.
4. Use this function to create series data (like list)
5. Custamizable indexing is allowed
6. We can chnage datatype manually
7. Mutable data structure

```python
s1 = pd.Series([1,2,3,4])
s1
```
